```python
def ricorsiva(n):
    print("ricorsiva: ", n)
    if n == 0:
        print("ricorsiva: fine")
        return
    ricorsiva(n - 1) # attesa
    # riparte da qui
    print("Fine ricorsiva: ", n)
    
# ricorsiva(3)

def fattoriale(n):
    if n == 0:
        print("fattoriale: fine")
        return 1
    print("fattoriale ", n)
    precedente = fattoriale(n-1) # in attesa
    return n * precedente

fattoriale(5)

# 5 * fattoriale(4) -> 5 * 24
# 4 * fattoriale(3) -> 4 * 6
# 3 * fattoriale(2) -> 3 * 2
# 2 * fattoriale(1) -> 2 * 1
# 1 * fattoriale(0) -> 1 * 1
# 1

import sys

# Python ah aun limite di profondità dello stack
print(sys.getrecursionlimit())


```
