```python
def ricorsiva(n):
    print("ricorsiva: ", n)
    if n == 0:
        print("ricorsiva: fine")
        return
    ricorsiva(n - 1) # attesa
    # riparte da qui
    print("Fine ricorsiva: ", n)
    
# ricorsiva(3)

def fattoriale(n):
    if n == 0:
        print("fattoriale: fine")
        return 1
    print("fattoriale ", n)
    precedente = fattoriale(n-1) # in attesa
    return n * precedente

fattoriale(5)

# 5 * fattoriale(4) -> 5 * 24
# 4 * fattoriale(3) -> 4 * 6
# 3 * fattoriale(2) -> 3 * 2
# 2 * fattoriale(1) -> 2 * 1
# 1 * fattoriale(0) -> 1 * 1
# 1

import sys

# Python ah aun limite di profondità dello stack
print(sys.getrecursionlimit())


```

```python
class ContoBancario:
    
    operazioniTotaliBanca = 0
    
    def __init__(self, saldo, intestatario):
        self.intestatario = intestatario
        self.saldo = saldo
        self.operazioniCorrentista = 0
        
    def info(self):
        print(f"Conto Bancario di {self.intestatario} saldo: €{self.saldo} operazioni eff:{self.operazioniCorrentista}/{self.operazioniTotaliBanca}")
    def deposita(self, importo):
        self.saldo += importo
        self.incrementa()
        self.operazioniCorrentista += 1
        
        
    def preleva(self, importo):
        if importo > self.saldo:
            print("Fondi insufficienti")
        else:
            self.saldo -= importo
            self.incrementa()
        self.operazioniCorrentista += 1
        
    @classmethod
    def incrementa(cls):
        cls.operazioniTotaliBanca += 1
        
        
conto1 = ContoBancario(100, "Mario Rossi")
conto1.deposita(350)
conto1.preleva(80)

conto2 = ContoBancario(250, "Giuseppe Verdi")
conto2.deposita(60)

conto1.info()
conto2.info()

print(f"Operazioni totali Banca: {ContoBancario.operazioniTotaliBanca}")
```

Gestione di un Corso Online

Implementare una classe Corso che rappresenti un corso online con studenti iscritti, gestione posti disponibili e validazioni.

Crea una classe: class Corso:

Il costruttore deve inizializzare:
nome (stringa)
posti_massimi (intero)
lista studenti (inizialmente vuota)

Vincoli:
nome non può essere vuoto
posti_massimi deve essere > 0

posti_disponibili calcolato come:
posti_massimi - numero_studenti

numero_studenti
lunghezza della lista studenti

Metodi di istanza
a) iscrivi_studente(nome_studente)
Aggiunge uno studente
❌ Non deve superare posti_massimi
❌ Non deve accettare duplicati
b) rimuovi_studente(nome_studente)
Rimuove uno studente
Se non esiste → messaggio di errore
c) elenco_studenti()
Restituisce lista studenti
d) descrizione()
Restituisce stringa tipo:
"Corso Python - 3/10 studenti iscritti"

Metodo di classe
a) crea_corso_base(cls, nome)
Crea un corso con:
posti_massimi = 10
Restituisce istanza

Aggiungere logging per:
✔ iscrizione studente
✔ rimozione studente
✔ errori (posti pieni, duplicati, ecc.)

Scrivere test per:
✔ creazione corso valida
✔ nome non valido → errore
✔ iscrizione studente
✔ iscrizione duplicata → errore
✔ superamento posti → errore
✔ rimozione studente
✔ rimozione studente inesistente → errore
✔ proprietà posti_disponibili
✔ metodo di classe

```python
import logging

logging.basicConfig(level=logging.DEBUG)

class Corso:
    def __init__(self, nome, posti_massimi):
        # Vincoli:
        # nome non può essere vuoto
        # posti_massimi deve essere > 0
        if not nome:
            logging.error('Nome non valido')
            raise ValueError('Nome non valido')
        if posti_massimi <= 0:
            logging.error('Posti massimi non validi')
            raise ValueError('Posti massimi non validi')
        
        self.nome = nome
        self.posti_massimi = posti_massimi
        self.studenti = []
    
    def numero_studenti(self):
        return len(self.studenti)
    
    def posti_disponibili(self):
        return self.posti_massimi - self.numero_studenti()
    
    def iscrivi_studente(self, nome_studente):
        # Non deve superare posti_massimi
        # Non deve accettare duplicati
        if not nome_studente:
            logging.error('Nome studente non valido')
            raise ValueError('Nome studente non valido')
        if nome_studente in self.studenti:
            logging.error('Studente già iscritto')
            raise ValueError('Studente già iscritto')
        if self.posti_disponibili() <= 0:
            logging.error('Posti terminati!')
            raise ValueError('Posti terminati!')
            
        self.studenti.append(nome_studente)
        logging.info('Studente iscritto al corso')
        
    def rimuovi_studente(self, nome_studente): 
        # Rimuove uno studente
        # Se non esiste → messaggio di errore
        if nome_studente not in self.studenti:
            logging.error('Nome studente non trovato')
            raise ValueError('Nome studente non trovato')
        
        self.studenti.remove(nome_studente)
        logging.info('Studente rimosso dal corso')
        
    def elenco_studenti(self):
        return list(self.studenti) # copia per sicurezza
    
    def descrizione(self):
        # Restituisce stringa tipo:
        # "Corso Python - 3/10 studenti iscritti"
        return f"Corso {self.nome} - {self.numero_studenti()}/{self.posti_massimi} studenti iscritti."
    
    @classmethod
    def crea_corso_base(cls, nome):
        # Crea un corso con: posti_massimi = 10
        # Restituisce istanza
        return cls(nome, 10)
    
    

python = Corso("Python", 3)
python.iscrivi_studente("Mario Rossi")
python.iscrivi_studente("Giuseppe Verdi")
python.iscrivi_studente("Francesca Neri")
# python.iscrivi_studente("")
# python.iscrivi_studente("Mario Rossi")
# python.iscrivi_studente("Antonio Bianchi")

python.descrizione()

python.rimuovi_studente("Giuseppe Verdi")
#python.rimuovi_studente("Antonio Bianchi")
    
python.elenco_studenti()

js = Corso.crea_corso_base("Javascript")
js.descrizione()
```
