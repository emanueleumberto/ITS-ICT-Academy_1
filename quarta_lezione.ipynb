```python
# Collection
# List | Tuple | Set | Dictionary
# Le collezioni list, set, tuple, dictionary posso essere create con valori di tipo diverso

# - List ->  Collezione di dati ORDINATA, MODIFICABILE e che permette valori DUPLICATI
#           v = ['Roma', 'Milano', 'Napoli']
# - Tuple -> Collezione di dati ORDINATA, IMMUTABILI e che permette valori DUPLICATI
#           v = ('Roma', 'Milano', 'Napoli')
# - Set ->   Collezione di dati NON ORDINATA, per questo NON INDICIZZABILE, 
#            NON MODIFICABILE e NON permette valori DUPLICATI
#           v = {'Roma', 'Milano', 'Napoli'}
# - Dictionary -> Collezione di dati ORDINATA e MODIFICABILE ma NON permette valori DUPLICATI
#           v = {"nome": "Mario", "cognome": "Rossi", "anni": 25}

# List
# Collezione di dati ORDINATA, MODIFICABILE e che permette valori DUPLICATI
# list() | type() | len() | count(val)
# è possibile accedere ad elementi di una lista tramite l'indice
# l[i] | l[i:i] | l[:i] | l[i:] | l[-i:-i]
# è possibile modificare elementi di una lista tramite l'indice
# l[i] = 'Nuovo valore' | l[i:i] = [val1, ..., valN]
# è possibile aggiungere degli elementi ad una lista
# append('val') | insert(i, val)
# è possibile rimuovere elementi da una lista
# remove(val) | pop() | pop(i) | del l[i] | del l | clear()
# è possibie unire due o più liste 
# extend(newList)
# è possibile ordinare i valori contenuti in una lista
# sort() | sort(reverse)
# è possibile copiare una lista 
# l.copy() | list(l)
# è possibile iterare una lista utilizzanmdo il ciclo FOR
# for ele in list:
#   blocco di istruzioni

#       0       1         2        3        -> indexing
l = ['Roma', 'Milano', 'Napoli', 'Roma']
#    |      |         |         |      |
#    0      1         2         3      4    -> slicing
#   -4      -3       -2         -1
print(l, type(l))
l = list(('Roma', 'Milano', 'Napoli', 'Roma'))
print(l, type(l))
print('Lunghezza della lista: ', len(l))
print('Numero di valori "Roma" nella lista: ', l.count('Roma'))
print(l[1])     # -> indexing
print(l[1:3])   # -> slicing
print(l[:3])
print(l[1:])
print(l[-3:-1])

l[1] = 'Palermo'
print(l)
l[1:3] = ['Torino', 'Firenze']
print(l)
l.append('Milano')
l.insert(1, 'Latina')
l.append('Torino')

# l.remove('Firenze')
# l.pop()
# l.pop(1)
# l.clear()
# del l
print(l)

l2 = ['Udine', 'Sezze', 'Viterbo']
l.extend(l2)
# l += l2
print(l)
l.sort()
print(l)
# l.sort(reverse=True)
# print(l)

nl = l # assegno la stessa zona di memoria di L a NL
nl = l.copy() # creo una copia di L e lo salvo nella variabile NL
nl = list(l) # creo una copia di L e lo salvo nella variabile NL

# nl.clear()
# print(nl)
# print(l)

def miaFunc(param):
    if param is not None :
        print(param)
        param.clear()
        
miaFunc(l)
print(l)

```

```python
# - Tuple -> Collezione di dati ORDINATA, IMMUTABILI e che permette valori DUPLICATI
# tuple() | type() | len() | count(val)
# è possibile accedere ad elementi di una tupla tramite l'indice
# t[i] | t[i:i] | tl[:i] | t[i:] | t[-i:-i]
# Non è possibile modificare una tupla
# è possibile fare l'unpack di una tupla 
# è possibie unire due o più tuple con l'operatore + 
# è possibile copiare una lista 
# tuple(t)
# è possibile trasformare una tupla in una lista
# è possibile iterare una lista utilizzanmdo il ciclo FOR
# for ele in list:
#   blocco di istruzioni


tp = ('Roma', 'Milano', 'Napoli')
print(tp, type(tp))
t = tuple(('Roma', 'Milano', 'Napoli', 'Firenze', 'Torino'))
print(t, type(t))

print(t[1])
print(t[:1])

# city1 = t[0]
# city2 = t[1]
# city3 = t[2]

# (city1, city2, city3, city4) = t # unpack di una collection
(city1, city2, *cities) = t # unpack di una collection

print(city1)
print(city2)
print(cities)

bigt = tp + t
print(bigt)

tcopy = tuple(bigt)

tcopyList = list(tcopy)
tcopyList.append('Palermo')
tcopy = tuple(tcopyList)
print(tcopy)
for ele in tcopy:
    print(ele)
```

```python
# - Set ->   Collezione di dati NON ORDINATA, per questo NON INDICIZZABILE, 
#            NON MODIFICABILE e NON permette valori DUPLICATI
#   set() | type() | len()
# Non possiamo modificare gli elementi di un set perchè non abbiamo un indice
# A differenza delle tuple però possiamo aggiungere e togliere degli elementi dal un set
# Per aggiungere elementi ad un set -> add(val) 
# Per rimuovere elementi da un set -> remove(val) | pop() | del s | clear()
# Unire set con union() -> Crea un nuovo set senza duplicati (Full Join)
# Unire set con intersection() -> Crea un nuovo set con i valori comuni tra i due set (Inner Join)
# Unire set con difference() -> Crea un nuovo set con valori univoci del set principale (Left Join) 
# Unire set con symmetric_difference() -> Crea un nuovo set con tutti gli elementi non comuni nei due set (Contrario di Inner Join)
# è possibile iterare una set utilizzando il ciclo FOR
# for ele in set:
#   blocco di istruzioni

s = {'Roma', 'Milano', 'Napoli'}
print(s, type(s))
st = set(('Roma', 'Milano', 'Napoli'))
print(st, type(st))

st.add('Torino')
st.add("Genova")
st.remove('Napoli')
st.pop()
st.clear()
# del st
print(st)

s1 = {'Roma', 'Genova', 'Pisa', 'Torino'}
s2 = {'Palermo', 'Roma', 'Torino', 'Firenze'}
s_union = s1.union(s2)
print(s_union)
s_intersection = s1.intersection(s2)
print(s_intersection)
s_difference = s1.difference(s2)
print(s_difference)
s_symmetric_difference = s1.symmetric_difference(s2)
print(s_symmetric_difference)


```

```python
# - Dictionary -> Collezione di dati ORDINATA e MODIFICABILE ma NON permette valori DUPLICATI
# I Dizionari sono un insieme di coppie chiave/valore, sono l'equivalente degli oggetti di altri linguaggi.
# dict() | type() | len()
# Per accedere ad un dizionario si utilizzano le chiavi -> d["chiave"]
# Per modificare un dizionario -> d["chiave"] = valore | d.update({"chiave": "valore"})
# Per rimuovere dati da un dizionario -> pop(chiave) | del d[chiave]
# Per copiare un dizionario -> d.copy() | dict(d)
# Per iterare un dizionario
# è possibile iterare un dizionari utilizzando il ciclo FOR
# for ele in dict: blocco di istruzioni
# for ele in dict.keys(): blocco di istruzioni
# for ele in dict.values(): blocco di istruzioni
# for ele in dict.items(): blocco di istruzioni

d = {"nome": "Mario", "cognome": "Rossi", "citta": "Roma", "anni": 25, "nome": "Antonio"}
print(d, type(d))
print(d["nome"])
d["cognome"] = "Verdi"
d.update({"citta": "Milano", "anni": 28})
# d.pop("citta")
# del d["anni"]
# d.clear()
# del d
print(d)
copyDict = d #Errato
copyDict = d.copy()
copyDict = dict(d)
print(copyDict)

for ele in copyDict: # itera le chivi di un dizionario
    print(ele, copyDict[ele])
    
for key in copyDict.keys(): # itera le chivi di un dizionario
    print(key, copyDict[key])

for val in copyDict.values(): # itera i valori di un dizionario
    print(val)
    
for item in copyDict.items(): # itera le coppie key/value di un dizionario
    print(item)
```
