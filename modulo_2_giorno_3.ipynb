La programmazione orientata agli oggetti (OOP) si basa su:
•	classi → modelli
•	oggetti → istanze
Permette di:
•	organizzare il codice
•	modellare il mondo reale
•	migliorare la riusabilità

Definizione di classe
class Persona:
    pass
Creazione oggetto:
p = Persona()

Metodi nelle classi
Un metodo è una funzione definita all’interno di una classe e serve a:
•	operare sui dati dell’oggetto
•	definire il comportamento dell’oggetto
•	incapsulare logica
Differenza fondamentale:
Funzione    ->	indipendente
Metodo      ->	legato a una classe

Metodo di istanza - È il tipo più comune.
Caratteristiche:
•	lavora sui dati dell’oggetto
•	riceve self come primo parametro

Il parametro self rappresenta:
•	l’istanza corrente
•	il collegamento tra metodo e oggetto

Metodi di classe
Un metodo di classe lavora sulla classe, non sull’istanza.
Si definisce con @classmethod.

Metodi statici
Non dipendono né da istanza né da classe.
Si definisce con @staticmethod.

Incapsulamento
È uno dei principi fondamentali della programmazione orientata agli oggetti (OOP)
Significa: nascondere i dati interni
permettere l’accesso solo tramite metodi controllati
Obiettivo: Proteggere lo stato interno di un oggetto
Evitare modifiche non valide
Rendere il codice più sicuro e manutenibile

Ereditarietà
L’ereditarietà permette a una classe di:
•	riutilizzare codice
•	estendere funzionalità
•	creare gerarchie

Polimorfismo
Il polimorfismo permette di usare oggetti diversi con la stessa interfaccia.

Ereditarietà multipla Python supporta più classi base.
class A: pass
class B: pass
class C(A, B): pass
⚠️ Problema:
•	complessità
•	ordine di risoluzione (MRO) Method Resolution Order
        Python decide quale metodo usare con l’MRO.



```python
class Veicolo:
    
    totale = 0
    
    # costruttore
    def __init__(self, marca, modello, targa):
        self.marca = marca
        self.modello = modello
        self.targa = targa
        self.incrementa()
    
    # Metodi di istanza
    def info(self):
        print(f"Veicolo: {self.marca} {self.modello} targa: {self.targa}")
        
    # Metodi di classe
    @classmethod
    def incrementa(cls):
        cls.totale += 1
    
    # Metodi static
    @staticmethod
    def test(t):
        return t * t
    
a = Veicolo("Fiat", "Panda", "AB123CD")
a.info()
Veicolo.incrementa()
a.incrementa()
Veicolo.test(20)
a.marca = "Ferrari"
```

```python
class Studente:
    
    def __init__(self, nome):
        self.nome = nome
        self.__voto = 0
    
    @property
    def voto(self):
        return self.__voto
    
    @voto.setter
    def voto(self, valore):
        if not isinstance(valore, (int, float)):
            raise TypeError("Il voto deve essere numerico")
        if valore < 0 or valore > 30:
            raise ValueError("Voto non valido (0-30)")

        self.__voto = valore 
        
    def is_promosso(self):
        return self.__voto >= 18       
    
    
s = Studente("Mario Rossi")
s.voto = 15           # setter
print(s.nome, s.voto) # getter
print("Promosso: ", s.is_promosso())
```

```python
# Ereditarietà
# Cos’è l’ereditarietà
# È un principio della programmazione orientata agli oggetti (OOP)
# Permette a una classe (figlia) di:
# riutilizzare codice di una classe (padre)
# estenderlo o modificarlo

class Veicolo:

    def __init__(self, marca, modello):
        self.__marca = marca
        self.__modello = modello
    
    def info(self):
        return f"Veicolo - {self.__marca} {self.__modello}"

class Automobile(Veicolo):
    
    # Uso di super() -> permette di richiamare il comportamento della classe padre.
    def __init__(self, marca, modello, porte):
        super().__init__(marca, modello)
        self.__porte = porte
        
        # Override (sovrascrittura)
        # Una classe figlia può modificare un comportamento del padre.
        
    def info(self):
        return f"{super().info()} Numero porte: {self.__porte}" 
     
class Moto(Veicolo):

    def __init__(self, marca, modello, posti):
        super().__init__(marca, modello)
        self.__posti = posti
        
    def info(self):
        return f"{super().info()} Numero posti: {self.__posti}" 
    
class Camper(Veicolo):
    
    def __init__(self, marca, modello, posti_letto):
        super().__init__(marca, modello)
        self.__posti_letto = posti_letto
        
    def info(self):
        return f"{super().info()} Numero posti letto: {self.__posti_letto}" 
    
# Errato -> Uno Smartphone NON è un Veicolo
# class Smartphone(Veicolo):
#    memoria = 0

m = Moto("Honda", "Hornet", 2)
a = Automobile("Renault", "Captur", 5)
c = Camper("Laika", "Camping", 7)

# Polimorfismo - permette di usare oggetti di tipo diverso con la stessa interfaccia
def leggi_info(veicolo):
    print(veicolo.info())

leggi_info(a)
leggi_info(m)
leggi_info(c)
```

```python
class Carrello:
    def __init__(self):
        self.prodotti = []
    
    def aggiungiProdotto(self, prezzo):
        self.prodotti.append(prezzo)    
        
    def __len__(self):
        return len(self.prodotti)
    
    def __str__(self):
        return f"Carrello con {len(self)} prodotti."
    
    def __add__(self, other):
        nuovoCarrello = Carrello()
        nuovoCarrello.prodotti = self.prodotti + other.prodotti
        return nuovoCarrello
    
c1 = Carrello()
c1.aggiungiProdotto(10)
c1.aggiungiProdotto(23)
c1.aggiungiProdotto(15)

len(c1)
print(c1)

c2 = Carrello()
c2.aggiungiProdotto(45)
c2.aggiungiProdotto(29)
print(c2)

c3 = c1 + c2
print(c3)
```

```python
# Confronti (__eq__)

class Vettore:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f"Vettore(x:{self.x}, y:{self.y})"
    
    def __add__(self, other):
        if not isinstance(other, Vettore):
            raise TypeError("Operazione non supportata")
        return Vettore(self.x + other.x, self.y + other.y)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    
    
v1 = Vettore(1,2)
v2 = Vettore(1,2)
print(v1)
print(v2)

v3 = v1 + v2
print(v3)

print(v1 == v2)
print(v1 == v3)
```

Estendi la classe Libro aggiungendo:
__str__ → restituisce una rappresentazione leggibile
__len__ → restituisce il numero di pagine
__eq__ → due libri sono uguali se titolo e autore coincidono
Testa i metodi creando più oggetti

Vincoli:
Non usare confronti diretti fuori dal metodo __eq__
Dimostrare l’uso di print(), len() e confronto ==

Output atteso:
print(libro1) → "Il libro '1984' di George Orwell ha 328 pagine"
len(libro1) → 328

```python
class Libro:
    def __init__(self, titolo, autore, pagine):
        self.__autore = autore
        self.__titolo = titolo
        self.__pagine = pagine
        
    @property
    def autore(self): return self.__autore
    
    @property
    def titolo(self): return self.__titolo
    
    @property 
    def pagine(self): return self.__pagine
        
    def __str__(self):
        # Rappresentazione leggibile
        return f"Il libro '{self.titolo}' di {self.autore} ha {self.pagine} pagine"
    
    def __len__(self):
        # Permette uso di len()
        return self.pagine
    
    def __eq__(self, other):
        # Controllo sicurezza tipo
        if not isinstance(other, Libro):
            return False
        # Confronto logico
        return self.titolo == other.titolo and self.autore == other.autore

    
# Creazione oggetti
libro1 = Libro("1984", "George Orwell", 328)
libro2 = Libro("Il Signore degli Anelli", "J.R.R. Tolkien", 1200)
libro3 = Libro("1984", "George Orwell", 300)

# Stampa risultati
print(libro1)   # __str__
print(libro2)   # __str__
print(libro3)   # __str__

print(len(libro1))         # __len__
print(len(libro2))         # __len__
print(len(libro3))         # __len__

print(libro1 == libro2)    # False
print(libro1 == libro3)    # True

```

```python
import math

class Forma:
    def area():
        # Metodo astratto
        raise NotImplementedError("Metodo da implementare nelle sottoclassi")
    
class Rettangolo(Forma):
    def __init__(self, base, altezza):
        self.__base = base
        self.__altezza = altezza
        
    @property
    def base(self): return self.__base
    @property
    def altezza(self): return self.__altezza
    
    def area(self):
        return self.base * self.altezza   
    
class Cerchio(Forma):
    def __init__(self, raggio):
        self.__raggio = raggio
        
    @property
    def raggio(self): return self.__raggio
    
    def area(self):
        return math.pi * self.raggio ** 2
    
def calcola_area(forme):
    for f in forme:
        print(f"Area: {f.area():.2f}")
        
forme = [
    Rettangolo(4,5),
    Cerchio(5)
]

calcola_area(forme)
```

Il concetto di astrazione è presente anche in Python ed è uno dei pilastri della programmazione orientata agli oggetti, anche se viene implementato in modo più flessibile rispetto a linguaggi come Java o C#.

L’astrazione consiste nel:
definire interfacce comuni
nascondere i dettagli implementativi
Permette di concentrarsi su cosa fa un oggetto e non come lo fa


Confronto tra approcci
abc
✔ rigoroso
✔ sicuro
❌ più verboso
Duck typing
✔ pythonico
✔ flessibile
❌ meno sicuro
NotImplementedError
✔ semplice
❌ nessun enforcement

```python
# Classi astratte con abc (approccio formale)
# Python fornisce il modulo built-in abc (Abstract Base Classes).

from abc import ABC, abstractmethod

class Forma(ABC):
    @abstractmethod
    def area():
        # Metodo astratto: nessuna implementazione
        pass
    
class Rettangolo(Forma):
    def __init__(self, base, altezza):
        self.__base = base
        self.__altezza = altezza
        
    @property
    def base(self): return self.__base
    @property
    def altezza(self): return self.__altezza
    
    def area(self):
        return self.base * self.altezza   
    
# f = Forma() # ❌ TypeError
r = Rettangolo(4,5)
```

```python
# Astrazione “informale” (duck typing)
# ✔ Concetto chiave
# Non importa il tipo → importa il comportamento
# Python spesso usa un approccio più dinamico:

class Rettangolo:
    def __init__(self, base, altezza):
        self.__base = base
        self.__altezza = altezza
        
    @property
    def base(self): return self.__base
    @property
    def altezza(self): return self.__altezza
    
    def area(self):
        return self.base * self.altezza   
    
class Cerchio:
    def __init__(self, raggio):
        self.__raggio = raggio
        
    @property
    def raggio(self): return self.__raggio
    
    def area(self):
        return math.pi * self.raggio ** 2
    
def calcola_area(forme):
    for f in forme:
        print(f"Area: {f.area():.2f}")
        
forme = [
    Rettangolo(4,5),
    Cerchio(5)
]

calcola_area(forme)
```

Concetto teorico: gestione di date e tempo
La gestione di date e orari è fondamentale per:
•	logging
•	sistemi distribuiti
•	analisi dati temporali
•	applicazioni web
Python fornisce il modulo standard: datetime
che consente di:
•	rappresentare date e orari
•	effettuare operazioni temporali
•	formattare e parsare date

Tipi principali del modulo datetime
date      ->	solo data (anno, mese, giorno)
time      ->	solo orario
datetime  ->	data + ora
timedelta ->	differenza tra date
tzinfo    ->	gestione timezone

```python
from datetime import date

data = date(2025, 12, 10)
today = date.today()

print(data)
print(today)
```

```python
from datetime import datetime

data = datetime(2025,12,10,9,30,0,0)
now = datetime.now()

print(data)
print(now)

print(now.year)
print(now.month)
print(now.day)
print(now.hour)

# %Y	anno
# %m	mese
# %d	giorno
# %H	ora
# %M	minuti
# %S	secondi

formatted = now.strftime("%d-%m-%Y %H:%M")
print(formatted)

# Convertire stringa -> datetime
data_str = "20-3-2026"
dt_str = datetime.strptime(data_str, "%d-%m-%Y")
print(dt_str)
print(dt_str.month)
```
