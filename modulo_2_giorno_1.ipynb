```python
# Debug con print

def somma_lista(lista):
    totale = 0
    for ele in lista:
        print("numero", ele, "totale", totale)
        totale =+ ele # Errore
        print("totale dopo operazione", totale)
    return totale

print(somma_lista([1,2,3,4,5]))

# Debug
# Errore: totale =+ ele
# Significato reale: totale = (+ele)
# ✔ Correzione: totale += ele
```

```python
# Uso di assert
# Le assertions servono a verificare condizioni
def dividi(x, y):
    assert y != 0, "Divisione per zero!!!"
    return x/y

print(dividi(25,5))
print(dividi(25,0))
```

```python
# Il modulo pdb (Python Debugger)
# Python include un debugger integrato: pdb
import pdb

def test():
    x = 10
    pdb.set_trace() # si ferma
    x += 5
    print(x)
    
# test()

# Quando il programma raggiunge set_trace() si ferma.
# n -> next (prossima riga)
# s -> step (entra in una funzione)
# c -> continue (continua)
# p var -> print variable (stampa il valore di una variabile)
# q -> quit (esci)

# Da Python 3.7: breakpoint() È equivalente a pdb.set_trace() ma più flessibile.
def test():
    x = 10
    breakpoint()
    y = x + 5
    print(y)
test()
```

```python
# Debugging con logging
# Il modulo logging è più potente di print().
# Il logging NON è locale al file 
# è globale a livello di processo condiviso tra moduli
import logging

logging.basicConfig(
    level=logging.ERROR,
    format="%(levelname)s - %(message)s",
    force=True
    )

print(logging.getLogger().level) # Valore numerico del log impostato

# I livelli hanno una priorità numerica:
# DEBUG    -> 10 ->	dettagli tecnici
# INFO	   -> 20 ->	informazioni generali
# WARNING  -> 30 ->	potenziale problema
# ERROR    -> 40 ->	errore
# CRITICAL -> 50 ->	errore grave

logging.debug("Messaggio di debug")
logging.info("Messaggio di info")
logging.error("messaggio di error")
```
